```python
from langchain_community.document_loaders import PyMuPDFLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
```

```python
file_path = "/home/farheens/Desktop/Order_Management_System/app/documents/product-inquriy.pdf"
```

```python
loader = PyMuPDFLoader(file_path=file_path)
```

```python
pages = loader.load()
```

```python
pages
```

```python
for page in pages:
    print(page.page_content)
    break
```

```python
print(pages[0].page_content)
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=570,
    chunk_overlap=50)
```

```python
docs = splitter.split_documents(pages)
```

```python
docs[0].page_content
```

```python
string = ""
l = []
for d in docs[0].page_content:
    if d != "\n":
        string += d
    else:
        l.append(string)
        string=""
l.append(string)
```
