```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
import re
import json
from pathlib import Path
```

```python
vector_store_dir = r'/home/farheens/Desktop/Order_Management_System/vector_store/faiss'
embedding_model = r'sentence-transformers/all-MiniLM-L6-v2'
```

```python
embedder  = HuggingFaceEmbeddings(
    model_name = embedding_model,
    model_kwargs={"device":"cpu"}
)
```

```python
base = embedding_model.split("/")[-1]
```

```python
embedded_name = re.sub(r"[^a-z0-9]+","-",base.lower()).strip("-")
embedded_name
```

```python
folder_path = f'{vector_store_dir}/{embedded_name}'
```

```python
folder_path
```

```python
type(folder_path)
```

```python
loaded_vectors = FAISS.load_local(
    folder_path= folder_path,
    embeddings=embedder,
    index_name="product-inquriy",
    allow_dangerous_deserialization=True
)
```

```python
results = loaded_vectors.similarity_search_with_score("smart tv",4)
```

```python
for result in results:
    print(result)
```

```python
chunks = []
```

```python
for result,d in results:
    result  = result.page_content.replace("\n"," ").replace("\u2022", "").strip()
    print("*"*50)

    chunks.append(" ".join(result.split()))
```
